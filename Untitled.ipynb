```python
import pandas as pd
from sqlalchemy import create_engine

# Create SQLAlchemy engine
engine = create_engine(
    "mysql+mysqlconnector://root:Root_123@127.0.0.1/ipl_analysis"
)

# Import Matches
matches = pd.read_csv("IPL_Project/Matches.csv")
matches.to_sql(
    "matches",
    con=engine,
    if_exists="replace",
    index=False
)
print("Matches imported successfully!")

# Import Deliveries
deliveries = pd.read_csv("IPL_Project/deliveries.csv")
deliveries.to_sql(
    "deliveries",
    con=engine,
    if_exists="replace",
    index=False
)
print("Deliveries imported successfully!")
```
